```python

import subprocess
import sys
import importlib

# Libraries to check
packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "scikit-learn": "sklearn",
    "joblib": "joblib",
    "gymnasium": "gymnasium",
    "streamlit": "streamlit"
}

print("CHECKING PYTHON LIBRARIES...\n")

for package, module_name in packages.items():
    try:
        # Check if the library is installed
        module = importlib.import_module(module_name)

        # Print installed version
        print(f"{package}: Installed (Version {module.__version__})")

    except ImportError:
        print(f"{package}: Not installed. Installing...")

        # Install missing package
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", package
        ])

        # Import newly installed package
        module = importlib.import_module(module_name)

        print(f"{package}: Successfully installed "
              f"(Version {module.__version__})")

print("\nLIBRARY CHECK COMPLETED!")
```

```python
# Calculate IQR fences for BMI
bmi_q1 = df["bmi"].quantile(0.25)
bmi_q3 = df["bmi"].quantile(0.75)
bmi_iqr = bmi_q3 - bmi_q1

outlier_col = "bmi"
lower = bmi_q1 - 1.5 * bmi_iqr
upper = bmi_q3 + 1.5 * bmi_iqr

print(f"{outlier_col} Q1: {bmi_q1:.2f}")
print(f"{outlier_col} Q3: {bmi_q3:.2f}")
print(f"{outlier_col} IQR: {bmi_iqr:.2f}")
print(f"{outlier_col} lower fence: {lower:.2f}")
print(f"{outlier_col} upper fence: {upper:.2f}")
print("BMI values below lower fence:", (df[outlier_col] < lower).sum())
print("BMI values above upper fence:", (df[outlier_col] > upper).sum())

# Calculate IQR fences for charges
charges_q1 = df["charges"].quantile(0.25)
charges_q3 = df["charges"].quantile(0.75)
charges_iqr = charges_q3 - charges_q1

charges_lower = charges_q1 - 1.5 * charges_iqr
charges_upper = charges_q3 + 1.5 * charges_iqr

print(f"\\nCharges lower fence: {charges_lower:.2f}")
print(f"Charges upper fence: {charges_upper:.2f}")
print("Charges below lower fence:", (df["charges"] < charges_lower).sum())
print("Charges above upper fence:", (df["charges"] > charges_upper).sum())
```
